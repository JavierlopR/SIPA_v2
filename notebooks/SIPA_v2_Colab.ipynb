## 🚀 Paso 3: Arrancar Streamlit y Generar un Enlace Público Seguro

Dado que Google Colab se ejecuta en servidores remotos de Google, no podemos abrir `localhost:8501` directamente en tu navegador personal. Para solucionarlo, utilizaremos **Cloudflare Tunnel (cloudflared)**. 

Este servicio creará de manera segura y automática una URL pública temporal (con terminación `.trycloudflare.com`) para que puedas ingresar y usar el dashboard de SIPA v2 desde cualquier dispositivo.

```python
# 1. Descargar e instalar Cloudflare Tunnel en la máquina virtual de Colab
!wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb
!dpkg -i cloudflared-linux-amd64.deb

# 2. Arrancar el servidor de Streamlit en segundo plano
import subprocess
subprocess.Popen(["streamlit", "run", "src/app.py", "--server.port", "8501", "--server.address", "0.0.0.0"])
print("⚡ Servidor Streamlit iniciado en segundo plano en el puerto 8501.")

# 3. Iniciar el túnel de Cloudflare
# Busca en el output inferior un enlace que termine en '.trycloudflare.com' y haz clic en él.
!cloudflared tunnel --url http://localhost:8501
```

💡 **Instrucciones importantes:**
1. Tras ejecutar la última celda, aparecerán varios logs de Cloudflare en la pantalla.
2. Busca la línea que dice algo parecido a: 
   `+--------------------------------------------------------------------------------------+
` 
   `|  Your quick Tunnel has been created! Visit it at:                                    |
` 
   `|  https://your-unique-subdomain.trycloudflare.com                                     |
` 
   `+--------------------------------------------------------------------------------------+
`
3. **Haz clic en el enlace `.trycloudflare.com`** y se abrirá la aplicación premium de SIPA v2 completamente funcional.
4. Para detener la aplicación, presiona el botón de **Detener Celda** (círculo de Stop) en Google Colab.
